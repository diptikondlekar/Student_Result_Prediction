```python
#Machine Learning For Data Science
#1.Import Important Libraries

import pandas as pd
import numpy as np

from sklearn.neighbors import KNeighborsClassifier    #algorithms KNN
from sklearn.model_selection import train_test_split  #Data training or testing split
from sklearn.metrics import accuracy_score            #Model performance
```

```python
#2.Import Dataset
import kagglehub

# Download latest version
path = kagglehub.dataset_download("jayantiprasad/student-exam-results-dataset")

print("Path to dataset files:", path)
```

```python
df =pd.read_csv(f'{path}/results.csv')
df.head()
```

```python
#3.Data Cleaning/PreProcessing
df.columns
```

```python
df=df.drop(columns=['Unnamed: 0','Div'])
```

```python
df.head(2)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
#4.Split column in input/output

df.head()
```

```python
df.columns
```

```python
X= df[['Hindi', 'English', 'Science', 'Maths', 'History', 'Geograpgy', 'Total']]
Y= df['Results']
```

```python
#5.Split data into training and testing

X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train
```

```python
#6.Create the model

model=KNeighborsClassifier(
    n_neighbors=5
)
```

```python
#7.Train the model
model.fit(X_train,Y_train)
```

```python
#8.Prediction

predict= model.predict(X_test)
predict
```

```python
com= pd.DataFrame({'Actual':Y_test,'Prediction':predict})
com
```

```python
#9.Evaluation and Accuracy
score= accuracy_score(Y_test,predict)
print('The Accuracy is',score*100)
```

```python
#10.Deployment
import pickle

pickle.dump(model,open('model.pkl','wb'))
print("PKL File Created")
```
